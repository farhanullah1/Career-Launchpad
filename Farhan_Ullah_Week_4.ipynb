```python
# Get the mode (most frequent value)
a = df["When do you prefer to study?"].mode()

# Fill missing values with the top mode
df["When do you prefer to study?"] = df["When do you prefer to study?"].fillna(a[0])
```

```python
import matplotlib.pyplot as plt

age_counts = df["What is your age?"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(age_counts.index, age_counts.values)

plt.title("Distribution of Age")
plt.xlabel("Age Groups")
plt.ylabel("Number of Students")

plt.show()
```

```python
age_counts = df["What is your age?"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    age_counts,
    labels=age_counts.index,
    autopct="%1.1f%%"
)

plt.title("Distribution of Age")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Clean column names to remove accidental whitespace
df.columns = df.columns.str.strip()

plt.figure(figsize=(8, 5))

plt.scatter(
    df["What is your age?"],
    df["How many languages do you speak fluently?"],
    marker="^"
)

plt.title("Age vs Languages Spoken")
plt.xlabel("What is your age?")
plt.ylabel("How many languages do you speak fluently?")

plt.grid(True)
plt.show()
```
